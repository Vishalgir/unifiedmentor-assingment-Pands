## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not. 
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here. 


```python
ut_list = ['Chandigarh', 'Delhi', 'Pondicherry']

def is_ut(city_name):
    if city_name in ut_list:
        return 'Yes'
    else:
        return 'No'
```

```python
d3['is_ut'] = d3['city'].apply(is_ut)

print(d3)

```

The temperatures mentioned in ‘Temperature’ column are mentioned in Celsius, we need another column which contains the same in Fahrenheit. 

HINT - 
- Define a function c_to_f which takes input temp in celsius and returns a value with temperature in Fahrenheit.
- To check: c_to_f(10) should return 50. 


```python
def c_to_f(c):
    return (c * 9/5) + 32
```

```python
# check function c_to_f(10)

print({c_to_f(10)})
```

```python
# apply function c_to_f to d3 to create a column 'temp_farenhiet'


d3['temp_farenhiet'] = d3['Temperature'].apply(c_to_f)
print(d3)
```

## Groupby

```python
# Create a dataframe using dictionary of your choice

# data = 


data = {
    'Item': ['Espresso', 'Latte', 'Cappuccino', 'Americano'],
    'Price': [2.50, 3.75, 3.50, 2.75],
    'In_Stock': [True, True, False, True]
}

inventory_df = pd.DataFrame(data)
print(inventory_df)
```

```python
# Use Groupby of single column with aggregate sum()
stock_sum = inventory_df.groupby('In_Stock')['Price'].sum()

print(stock_sum)

```

```python
# Use Groupby of single column with aggregate count()

stock_count = inventory_df.groupby('In_Stock')['Item'].count()

print(stock_count)
```

```python
# Use Groupby of single column with aggregate min() and max()


stock_min_max = inventory_df.groupby('In_Stock')['Price'].agg(['min', 'max'])

print(stock_min_max)
```

```python
# Use Groupby of any 2 columns with aggregate mean()


multi_group_mean = inventory_df.groupby(['In_Stock', 'Item'])['Price'].mean()

print(multi_group_mean)
```

```python

data = {
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Clothing', 'Clothing', 'Clothing'],
    'Store_Location': ['North', 'South', 'North', 'North', 'South', 'North'],
    'Price': [500, 700, 450, 25, 40, 30]
}

df_sales = pd.DataFrame(data)


grouped_stats = df_sales.groupby(['Category', 'Store_Location'])['Price'].agg(['min', 'max'])

print(grouped_stats)
```
